```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_excel('Adidas US Sales Datasets.xlsx')

# Preview
print(df.head())
```

```python
total_sales = df["Total Sales"].sum()
units_sold = df["Units Sold"].sum()
avg_price = df["Price per Unit"].mean()

print("Total Sales:", total_sales)
print("Units Sold:", units_sold)
print("Avg Price per Unit:", avg_price)
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Units Sold", y="Price per Unit", hue="Product")

plt.title("Product vs Price")
plt.xlabel("Units Sold")
plt.ylabel("Price per Unit")
plt.show()
```

```python
state_retailer = df.groupby("Retailer")["State"].count()

plt.figure(figsize=(6,6))
plt.pie(state_retailer, labels=state_retailer.index, autopct='%1.1f%%')
plt.title("Count of State by Retailer")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Operating Margin", y="Operating Profit", hue="Retailer")

plt.title("Operating Profit vs Operating Margin")
plt.xlabel("Operating Margin")
plt.ylabel("Operating Profit")
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales Method"].count()

plt.figure(figsize=(8,5))
region_sales.plot(kind='bar')

plt.title("Count of Sales Method by Region")
plt.xlabel("Region")
plt.ylabel("Count")
plt.show()
```

```python
city_invoice = df.groupby("City")["Invoice Date"].count()

plt.figure(figsize=(10,5))
city_invoice.plot(kind='line')

plt.title("Count of Invoice Date by City")
plt.xlabel("City")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
sales_state = df.groupby("State").agg({
    "Total Sales": "sum",
    "State": "count"
}).rename(columns={"State": "Count of State"})

fig, ax1 = plt.subplots(figsize=(10,5))

# Bar chart
ax1.bar(sales_state.index, sales_state["Total Sales"])
ax1.set_ylabel("Total Sales")

# Scatter
ax2 = ax1.twinx()
ax2.scatter(sales_state.index, sales_state["Count of State"])
ax2.set_ylabel("Count of State")

plt.title("Total Sales and Count of State")
plt.xticks(rotation=45)
plt.show()
```
